```python
print("Loading model...")
model = LLM(
    model="Qwen/Qwen3-Embedding-8B"
)
print("Model loaded successfully!")
```

```python
import requests
import pandas as pd
import time

def github_dir_to_raw_skill(url):
    """
    Convert GitHub directory URL to raw SKILL.md URL
    
    Input:  https://github.com/user/repo/tree/branch/path/skill-name
    Output: https://raw.githubusercontent.com/user/repo/branch/path/skill-name/SKILL.md
    """
    raw = url.replace(
        "https://github.com/",
        "https://raw.githubusercontent.com/"
    ).replace("/tree/", "/")
    return f"{raw}/SKILL.md"

def fetch_content(github_dir_url, retries=3):
    """Fetch SKILL.md content from GitHub directory URL"""
    candidates = [
        github_dir_to_raw_skill(github_dir_url),
        github_dir_to_raw_skill(github_dir_url).replace("SKILL.md", "skill.md"),
    ]
    
    for raw_url in candidates:
        for attempt in range(retries):
            try:
                response = requests.get(raw_url, timeout=10)
                if response.status_code == 200:
                    return response.text, None
                elif response.status_code == 404:
                    break  # try next candidate
                elif response.status_code == 403:
                    return None, "rate_limited"
                else:
                    time.sleep(2)
            except Exception as e:
                time.sleep(2)
    
    return None, "not_found"

def clean_content(text):
    if not text:
        return None
    lines = [line for line in text.splitlines() if line.strip()]
    return '\n'.join(lines).strip()

# Load data
df = pd.read_csv("skillsmp_labeled_full_data.csv")
df_se = df[df['se_label'] == 'SE-related'].reset_index(drop=True)
print(f"SE-related skills: {len(df_se)}")

# Test on first 3 rows before full run
print("\n--- Testing first 3 rows ---")
for i in range(3):
    url = df_se['githubUrl'].iloc[i]
    content, error = fetch_content(url)
    print(f"[{i+1}] {url}")
    print(f"     Error: {error}")
    print(f"     Content: {content[:100] if content else 'None'}")
    print()
```

```python
from concurrent.futures import ThreadPoolExecutor, as_completed

def fetch_row(row):
    content, error = fetch_content(row['githubUrl'])
    return {
        'id': row['id'],
        'name': row['name'],
        'author': row['author'],
        'githubUrl': row['githubUrl'],
        'stars': row['stars'],
        'se_label': row['se_label'],
        'raw_content': content,
        'content_clean': clean_content(content),
        'content_length': len(content) if content else 0,
        'error': error
    }

results = []
with ThreadPoolExecutor(max_workers=10) as executor:
    futures = {
        executor.submit(fetch_row, row): i 
        for i, row in df_se.iterrows()
    }
    for future in as_completed(futures):
        results.append(future.result())
        if len(results) % 1000 == 0:
            print(f"Fetched: {len(results)}/{len(df_se)}")

df_results = pd.DataFrame(results)
df_results.to_csv("skills_with_content.csv", index=False)
```

```python
def is_valid_skill(content):
    if not content or len(content.strip()) < 50:
        return False, "too_short"
    
    # Check start for HTML
    start = content.strip()[:200].lower()
    if start.startswith("<html") or start.startswith("<!doctype"):
        return False, "html_page"
    
    has_yaml = content.strip().startswith("---")
    
    if has_yaml:
        # Find end of YAML block
        yaml_end = content.find("---", 3)  # find closing ---
        yaml_block = content[:yaml_end] if yaml_end > 0 else content[:500]
        has_name = "name:" in yaml_block
    else:
        has_name = "name:" in content[:500]
    
    if not has_yaml and not has_name:
        return False, "no_skill_structure"
    
    return True, None
    
# Apply to df_en
print("Validating skill content...")
results = df_en['content_clean'].apply(lambda x: is_valid_skill(x))
df_en['is_valid'] = results.apply(lambda x: x[0])
df_en['invalid_reason'] = results.apply(lambda x: x[1])

print(f"Valid skills: {df_en['is_valid'].sum()}")
print(f"Invalid skills: {(~df_en['is_valid']).sum()}")
print(f"\nInvalid reasons:")
print(df_en['invalid_reason'].value_counts())
```

```python
#%%
# Run this cell instead of section 6 on subsequent runs
import numpy as np
import pandas as pd

embeddings_norm = np.load("skill_embeddings_norm.npy")
df_en = pd.read_csv("skills_clean.csv")

skill_ids = df_en['id'].tolist()
skill_names = df_en['name'].tolist()

print(f"Loaded {len(skill_ids)} embeddings")
print(f"Shape: {embeddings_norm.shape}")
```

```python
#%%
import faiss

dim = embeddings_norm.shape[1]
print(f"Embedding dimension: {dim}")

# Build exact FAISS index
print("Building FAISS IndexFlatIP (exact cosine similarity)...")
index = faiss.IndexFlatIP(dim)

# Try GPU first, fall back to CPU
try:
    res = faiss.StandardGpuResources()
    gpu_index = faiss.index_cpu_to_gpu(res, 0, index)
    gpu_index.add(embeddings_norm)
    search_index = gpu_index
    print(f"Using GPU index with {search_index.ntotal} vectors")
except Exception as e:
    print(f"GPU not available ({e}), using CPU")
    index.add(embeddings_norm)
    search_index = index
    print(f"Using CPU index with {search_index.ntotal} vectors")
```

```python
#%%
# Search top-k similar skills
k = 50           # top 50 similar skills per skill
THRESHOLD = 0.7  # minimum cosine similarity

print(f"Searching top-{k} neighbors for {len(skill_ids)} skills...")
distances, indices = search_index.search(embeddings_norm, k)
# distances = cosine similarity scores (higher = more similar)
# indices = indices of similar skills
print("Search complete!")
```

```python
#%%
# Collect pairs above threshold
print(f"Collecting pairs with cosine similarity >= {THRESHOLD}...")
duplicates = []
n = len(skill_ids)

for i in range(n):
    for j_rank in range(k):
        j = int(indices[i, j_rank])
        sim = float(distances[i, j_rank])

        if j <= i:        # skip self-pairs and lower triangle
            continue
        if sim < THRESHOLD:
            break         # results sorted by similarity, safe to stop

        duplicates.append({
            'skill_1_id': skill_ids[i],
            'skill_2_id': skill_ids[j],
            'skill_1_name': skill_names[i],
            'skill_2_name': skill_names[j],
            'similarity': sim,
        })

    if i % 10000 == 0:
        print(f"Progress: {i}/{n} | Pairs found: {len(duplicates)}")

print(f"\nTotal pairs found: {len(duplicates)}")
```

```python
#%%
# Sort and save
df_sims = pd.DataFrame(duplicates)
df_sims = df_sims.sort_values('similarity', ascending=False).reset_index(drop=True)
df_sims.to_csv("skill_similarities_07.csv", index=False)
print(f"Saved {len(df_sims)} pairs to skill_similarities_07.csv")
```
